```python
import re

class EnhancedKhojaStemmer:
    def __init__(self):
        # القوائم الأساسية المحسنة
        self.prefixes = ["ال", "وال", "بال", "كال", "فال", "لل", "و", "ف", "ب", "ك", "ل", "س"]
        self.suffixes = ["ون", "ين", "ان", "ات", "وا", "تم", "كن", "نا", "ت", "ه", "ها", "هم", "هن", "ي"]
        
        # قائمة الأوزان (أمثلة)
        self.patterns = ["فعل", "فاعل", "مفعول", "افعل", "تفاعل", "استفعل", "فعال"]

    def clean_text(self, text):
        # تنظيف النص من التشكيل والرموز
        text = re.sub(r'[ًٌٍَُِّْ]', '', text)
        return text

    def remove_prefixes(self, word):
        for p in sorted(self.prefixes, key=len, reverse=True):
            if word.startswith(p) and len(word) > len(p) + 2:
                word = word[len(p):]
                break
        return word

    def remove_suffixes(self, word):
        for s in sorted(self.suffixes, key=len, reverse=True):
            if word.endswith(s) and len(word) > len(s) + 2:
                word = word[:-len(s)]
                break
        return word

    def stem(self, word):
        word = self.clean_text(word)
        # تطبيق الخطوات بترتيب منطقي
        stemmed = self.remove_prefixes(word)
        stemmed = self.remove_suffixes(stemmed)
        
        # إضافة منطق التحقق من الأوزان هنا (اختياري)
        return stemmed

# تجربة الكود المحسن
stemmer = EnhancedKhojaStemmer()
words = ["المجاهدون", "بالكتاب", "سيذهبون", "الشجرات"]
for w in words:
    print(f"الكلمة: {w} -> الجذر المحسن: {stemmer.stem(w)}")
```

```python
# 1. ترتيب الحذف (Longest Match First)
# في الدوال remove_prefixes و remove_suffixes استخدمنا: sorted(self.prefixes, key=len, reverse=True)

# التحسين: الخوارزمية الآن تحذف السوابق الأطول أولاً.

# مثال: كلمة "بالكتاب"، إذا بحثت الخوارزمية عن "بـ" أولاً ستحذفها وتبقى "الكتاب"، لكن التحسين يجعلها تبحث عن "بالـ" كقطعة واحدة، مما يقلل عدد العمليات ويمنع الأخطاء في فصل الحروف.

# 2. شرط طول الكلمة المتبقية (Length Constraint)
# أضفنا شرط: if ... and len(word) > len(p) + 2

# التحسين: هذا يمنع الخوارزمية من حذف حروف أصلية من الكلمة.

# الفائدة: خوارزمية خوجة الأصلية قد تحول كلمة "علم" إلى "ع" إذا اعتبرت "لم" لاحقة. هذا الشرط يضمن أن الكلمة بعد الحذف لا يقل طولها عن حرفين أو ثلاثة (جذر منطقي).

# 3. شمولية السوابق واللواحق
# التحسين: تمت إضافة سوابق مركبة مثل (فال، وال، كال، لل) ولواحق مثل (تم، كن، هن).

# الفائدة: خوجة التقليدية كانت تركز على الحروف المفردة (أ، و، ي، ت، ن، ل، ب، ك، س، م). هذه القائمة المحدثة تتعامل مع التعقيد الحديث للنصوص.
```
