```python
from camel_tools.morphology.database import MorphologyDB
from camel_tools.morphology.analyzer import Analyzer
import pandas as pd

# 1. إعداد المحلل الصرفي (استخدام قاعدة البيانات الافتراضية)
db = MorphologyDB.builtin_db()
analyzer = Analyzer(db)

def analyze_with_camel(text_list):
    results = []
    
    for word in text_list:
        # التحليل الصرفي يعطي قائمة من الاحتمالات، سنأخذ الاحتمال الأول الأرجح
        analyses = analyzer.analyze(word)
        
        if len(analyses) > 0:
            top_analysis = analyses[0] # الاحتمال الأعلى دقة
            results.append({
                'الكلمة': word,
                'الجذر (Root)': top_analysis.get('root', 'N/A'),
                'النمط (Pattern)': top_analysis.get('pattern', 'N/A'),
                'التشكيل': top_analysis.get('diac', 'N/A'),
                'نوع الكلام': top_analysis.get('pos', 'N/A')
            })
        else:
            results.append({'الكلمة': word, 'الجذر (Root)': 'لم يتم العثور'})

    return pd.DataFrame(results)

# 2. تنفيذ التحليل على كلماتك
words_to_test = ["المسافرون", "يتحدثون", "بالجامعات", "استخراج"]
df_camel = analyze_with_camel(words_to_test)

# 3. عرض النتائج بشكل جدول منسق
print("--- نتائج تحليل خوارزمية CAMel Tools ---")
print(df_camel.to_markdown(index=False))
```
