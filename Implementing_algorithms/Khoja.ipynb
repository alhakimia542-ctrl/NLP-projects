```python
from datasets import load_dataset

# تحميل Dataset الجذور
dataset = load_dataset("MohamedRashad/arabic-roots")

# تحويل الجذور إلى set (سريع جدًا في البحث)
ROOTS = set(dataset["train"]["root"])

print("عدد الجذور:", len(ROOTS))
```

```python
import re

def normalize(word):
    # إزالة التشكيل
    word = re.sub(r"[ًٌٍَُِّْـ]", "", word)

    # توحيد الهمزات
    word = re.sub(r"[أإآ]", "ا", word)
    word = re.sub(r"[ؤئ]", "ء", word)

    # إزالة التطويل
    word = re.sub(r"ـ", "", word)

    return word
```

```python
PREFIXES = [
    "ال", "وال", "بال", "كال", "فال", "لل",
    "و", "ف", "ب", "ك", "ل", "س"
]

SUFFIXES = [
    "ه", "ها", "هم", "هن", "كما", "كم", "كن",
    "نا", "ي", "ة", "ات", "ان", "ون", "ين"
]

def strip_affixes(word):
    for p in PREFIXES:
        if word.startswith(p):
            word = word[len(p):]

    for s in SUFFIXES:
        if word.endswith(s):
            word = word[:-len(s)]

    return word
```

```python
PATTERNS = {
    "مفعول": (1, 2, 3),
    "فاعل": (0, 2, 3),
    "تفعيل": (1, 2, 3),
    "افتعال": (1, 3, 4),
    "استفعال": (2, 4, 5),
    "انفعال": (2, 3, 4),
    "فعالة": (0, 1, 2),
    "فعول": (0, 1, 2),
    "فعيل": (0, 1, 2),
}
```

```python
def extract_root_by_pattern(word):
    for pattern, idx in PATTERNS.items():
        if len(word) == len(pattern):
            root = "".join(word[i] for i in idx)
            if root in ROOTS:
                return root
    return None
```

```python
def khoja_stem(word):
    word = normalize(word)
    word = strip_affixes(word)

    # إذا أصبحت الكلمة قصيرة جدًا
    if len(word) < 3:
        return word

    # 1️⃣ إذا الكلمة نفسها جذر
    if word in ROOTS:
        return word

    # 2️⃣ محاولة استخراج الجذر بالأنماط
    root = extract_root_by_pattern(word)
    if root:
        return root

    # 3️⃣ fallback: أول 3 أحرف
    guess = word[:3]
    if guess in ROOTS:
        return guess

    return word
```

```python
from datasets import load_dataset
import re

dataset = load_dataset("MohamedRashad/arabic-roots")
roots = dataset["train"]["root"]

print("عدد الجذور الكلي:", len(roots))
```

```python
unique_roots = set(roots)

print("عدد الجذور الفريدة:", len(unique_roots))
print("عدد المكررات:", len(roots) - len(unique_roots))
```

```python
from collections import Counter

lengths = [len(r) for r in roots]
length_stats = Counter(lengths)

print("توزيع أطوال الجذور:")
for length, count in sorted(length_stats.items()):
    print(f"طول {length} : {count}")
```

```python
def is_valid_arabic_root(root):
    return bool(re.fullmatch(r"[ء-ي]+", root))

invalid_roots = [r for r in roots if not is_valid_arabic_root(r)]

print("عدد الجذور غير القياسية:", len(invalid_roots))
print("أمثلة:")
print(invalid_roots[:10])
```

```python
short_roots = [r for r in roots if len(r) < 3]
long_roots = [r for r in roots if len(r) > 4]

print("جذور قصيرة جدًا (<3):", len(short_roots))
print("جذور طويلة جدًا (>4):", len(long_roots))
```

```python
print("===== DATASET QUALITY REPORT =====")
print("Total roots:", len(roots))
print("Unique roots:", len(unique_roots))
print("Duplicates:", len(roots) - len(unique_roots))
print("Invalid roots:", len(invalid_roots))
print("Short roots:", len(short_roots))
print("Long roots:", len(long_roots))
print("=================================")
```
