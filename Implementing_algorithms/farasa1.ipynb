```python
import pandas as pd
from farasa.segmenter import FarasaSegmenter
from farasa.tagger import FarasaPOSTagger
from farasa.stemmer import FarasaStemmer
import os

def run_comprehensive_analysis():
    # 1. إدخال النص من المستخدم
    print("--- محرك معالجة اللغة العربية (فراسة) ---")
    input_text = input("أدخل الجملة التي تريد تحليلها: ") or "سيذهب الطلاب إلى المدرسة مبكرين"

    try:
        # 2. تهيئة الأدوات (سيقوم بتحميل ملفات الـ JAR تلقائياً في أول مرة)
        print("\n[1/3] جاري تحضير المحللات... قد يستغرق ذلك لحظات في المرة الأولى.")
        segmenter = FarasaSegmenter(interactive=True)
        tagger = FarasaPOSTagger(interactive=True)
        stemmer = FarasaStemmer(interactive=True)

        # 3. استخراج البيانات
        print("[2/3] جاري معالجة النص...")
        
        # التقطيع (Segmentation)
        segmented_words = segmenter.segment(input_text).split()
        
        # التوسيم (POS Tagging) - يعيد قائمة من القواميس تحتوي على word و part
        tags = tagger.tag(input_text)
        
        # استخراج الجذور (Stemming)
        stems = stemmer.stem(input_text).split()

        # 4. تجميع البيانات في جدول
        # ملاحظة: نقوم بضبط الأطوال لضمان توافق الجدول
        max_len = len(segmented_words)
        table_data = []

        for i in range(max_len):
            word_tag = tags[i] if i < len(tags) else {'word': '?', 'part': '?'}
            table_data.append({
                "الكلمة الأصلية/المقطعة": segmented_words[i],
                "الجذر (Stem)": stems[i] if i < len(stems) else "غير متوفر",
                "النوع (POS Tag)": word_tag['part'],
                "التفاصيل": word_tag['word']
            })

        # 5. عرض النتائج باستخدام Pandas
        df = pd.DataFrame(table_data)
        
        print("\n[3/3] اكتمل التحليل! إليك النتائج:")
        print("-" * 50)
        # استخدام tabulate لتحسين المظهر في الـ Console
        print(df.to_string(index=False, justify='center'))
        print("-" * 50)

    except Exception as e:
        print(f"\nحدث خطأ: {e}")
        print("تأكد من تثبيت Java (JRE) على جهازك وإضافته للمسار (PATH).")

if __name__ == "__main__":
    run_comprehensive_analysis()
```
