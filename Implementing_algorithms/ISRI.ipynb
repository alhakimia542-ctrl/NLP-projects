```python
from datasets import load_dataset

dataset = load_dataset("wiki40b", "ar", split="train[:1%]")

print("عدد النصوص:", len(dataset))
print(dataset[0]["text"][:500])  # عرض أول 500 حرف من النص الأول
```

```python
from nltk.stem.isri import ISRIStemmer

stemmer = ISRIStemmer()
```

```python
def stem_text(text):
    words = text.split()
    stems = [stemmer.stem(w) for w in words]
    return stems
```

```python
first_text = dataset[0]["text"]
first_stems = stem_text(first_text)

print("أول 20 كلمة أصلية:", first_text.split()[:20])
print("أول 20 جذر مستخرج:", first_stems[:20])
```

```python
all_stems = []

for text in dataset["text"]:
    stems = stem_text(text)
    all_stems.append(stems)

print("تم استخراج الجذور لجميع النصوص!")
print("مثال على أول نص:", all_stems[0][:20])
```
