```python
# نصوص بسيطة
documents = [
    "I love NLP",
    "NLP is fun",
    "I love programming"
]

# 1️⃣ جمع كل الكلمات الفريدة
vocabulary = []
for doc in documents:
    for word in doc.lower().split():
        if word not in vocabulary:
            vocabulary.append(word)

print("Vocabulary:", vocabulary)

# 2️⃣ إنشاء مصفوفة BoW
for doc in documents:
    vector = []
    words = doc.lower().split()
    for word in vocabulary:
        vector.append(words.count(word))  # عد مرات ظهور الكلمة
    print(f"{doc} → {vector}")
```
