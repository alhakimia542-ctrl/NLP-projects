```python
from PyPDF2 import PdfReader

reader = PdfReader("Arduino.pdf")

print("عدد الصفحات:", len(reader.pages))

text = reader.pages[0].extract_text()
print(text)
```

```python
from reportlab.platypus import SimpleDocTemplate, Paragraph
from reportlab.lib.styles import getSampleStyleSheet

doc = SimpleDocTemplate("arduino_output.pdf")
styles = getSampleStyleSheet()

paragraphs = []
for line in text.split("\n"):
    paragraphs.append(Paragraph(line, styles["Normal"]))

doc.build(paragraphs)
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas

c = canvas.Canvas("new_file.pdf", pagesize=A4)

c.drawString(100, 800, "مرحبا، هذا ملف PDF جديد")
c.drawString(100, 770, "تم إنشاؤه باستخدام بايثون")

c.save()
```

```python
from pypdf import PdfReader, PdfWriter

# قراءة الملف الأصلي
reader = PdfReader("Arduino.pdf")

# إنشاء كاتب PDF جديد
writer = PdfWriter()

# إضافة أول صفحة فقط
first_page = reader.pages[0]
writer.add_page(first_page)

# حفظ الصفحة في ملف جديد
with open("first_page.pdf", "wb") as f:
    writer.write(f)
```
